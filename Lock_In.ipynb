EXAMPLE : $$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

```python
# Level 1 : raw loop
def calculate_mse(y, y_hat):
    n = len(y)
    total_error = 0.0
    for i in range(n):
        diff = y[i] - y_hat[i]
        total_error += diff ** 2
    return total_error / n
```

```python
# level 2 : Tư duy mảng (Vectorization - NumPy/Tensor)
import numpy as np

def calculate_mse_vec(y, y_hat):
    # Phép trừ và mũ hai thực hiện song song trên từng phần tử
    return np.mean((y - y_hat) ** 2)
```

### 3.khoảng cách Euclidean & Manhattan giữa 2 vector $n$ chiều:$$d_{\text{Euclid}}(u, v) = \sqrt{\sum_{i=1}^n (u_i - v_i)^2}, \quad d_{\text{Manhattan}}(u, v) = \sum_{i=1}^n \vert{}u_i - v_i\vert{}$$

```python
# euclidean level 1
def caculate_euclidean(u,v):
  n = len(u)
  total = 0.0
  for i in range(n):
    diff = u[i] - v[i]
    total += diff**2
  return np.sqrt(total)
```

```python
# manhattan level 1
def caculate_manhattan(u,v):
  n = len(u)
  total = 0.0
  for i in range(n):
    diff = u[i] - v[i]
    total += np.squrt(diff**2)
  return total
```

### 4. Tích vô hướng (Dot Product) và Cosine Similarity:$$\text{Cosine}(u, v) = \frac{u \cdot v}{\Vert{}u\Vert{} \Vert{}v\Vert{}} = \frac{\sum u_i v_i}{\sqrt{\sum u_i^2} \sqrt{\sum v_i^2}}$$

```python
# Dot product & cosine similarity level 1
def caculate_ds1(u,v):
  n = len(u)
  total_1 = 0.0
  total_2 =0.0
  total_3 = 0.0
  total =0.0
  for i in range(n):
    diff_1 = u[i] * v[i]
    diff_2 = np.squrt(u[i]**2)
    diff_3 = np.squrt(v[i]**2)
    total_1 += diff_1
    total_2 += diff_2
    total_3 += diff_3
    total = total_1 / (total_2 * total_3)

  return total

```
