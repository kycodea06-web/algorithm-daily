```python
# Level 1 : raw loop
def calculate_mse(y, y_hat):
    n = len(y)
    total_error = 0.0
    for i in range(n):
        diff = y[i] - y_hat[i]
        total_error += diff ** 2
    return total_error / n
```

```python
# Dot product & cosine similarity level 1
def caculate_ds1(u,v):
  n = len(u)
  total_1 = 0.0
  total_2 =0.0
  total_3 = 0.0
  total =0.0
  for i in range(n):
    diff_1 = u[i] * v[i]
    diff_2 = np.squrt(u[i]**2)
    diff_3 = np.squrt(v[i]**2)
    total_1 += diff_1
    total_2 += diff_2
    total_3 += diff_3
    total = total_1 / (total_2 * total_3)

  return total
```

## Cấp độ 2: Hàm kích hoạt & Loss Functions trong ML

### Softmax: $\sigma(z)_i = \frac{e^{z_i}}{\sum_{j=1}^K e^{z_j}}$ (chú ý mẹo trừ max(z) để tránh tràn số numerical overflow)

```python
#Level 1
import math
def soft_max(z):
  # Bước 1: Tìm giá trị lớn nhất trong mảng z
  max_val = max(z)
  # Bước 2: Trừ max_val cho từng phần tử và tính e^(z_i - max_val)
  exp_values = [math.exp(x - max_val) for x in z]
  # Bước 3: Tính tổng mẫu số
  sum_exp = sum(exp_values)
  # Bước 4: Chia từng phần tử cho tổng để ra xác suất
  probabilities = [val / sum_exp for val in exp_values]
  return probabilities
```

```python
# Level 2
def soft_max_vec(z):
  exp_values = np.exp(z-np.max(z))
  probabilities = exp_values / np.sum(exp_values)
  return probabilities
```

### Binary Cross-Entropy Loss: $\mathcal{L} = -\frac{1}{N} \sum_{i=1}^N \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$

```python
#level 1
def binary_cross_entropy(y_true, y_pred):
  n = len(y_true)
  total = 0.0
  for i in range(n):
    diff = y_true[i]*np.log(y_pred[i]) + (1-y_pred)*np.log(1-y_pred[i])
    total += diff
  return -total/n
```

```python
#level 2
def binary_cross_entrophy_vec(y_true, y_pred):
  return -np.mean(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))
```

### Min-Max Normalization: $x' = \frac{x - \min(x)}{\max(x) - \min(x)}$

```python
#level 1
def min_max(x):
  x_min = min(x)
  x_max = max(x)
  x_norm = [(x[i] - x_min)/(x_max - x_min) for i in range(len(x))]
  return x_norm
```

```python
#level 2
def min_max_vec(x):
  return (x - np.min(x))/(np.max(x) - np.min(x))
```

## Cấp độ 3: Công thức cập nhật lặp (Optimization / State Updates)

### Luyện công thức có tham số thời gian/bước lặp $t$:

#### Gradient Descent: $\theta^{(t+1)} = \theta^{(t)} - \alpha \nabla J(\theta^{(t)})$

```python
# Level 1
# Định nghĩa hàm chi phí và đạo hàm
def cost_function(theta):
  return theta**2 - 4 * theta +3
def compute_gradient(theta):
  return 2 * theta - 4

# Thiết lập siêu tham số
learning_rate = 0.1
num_interations = 30 # Số vòng lặp tối đa
tolerance = 1e-6 # Sửa lỗi cú pháp: viết liền 1e-6 để làm ngưỡng dừng

# Khởi tạo giá trị ban đầu theta^(0)
theta = 10.0
print(f"Bắt đầu: theta = {theta:.4f}, Cost = {cost_function(theta):.4f}\n")

# 4. Vòng lặp cập nhật Gradient Descent
for i in range(num_interations):
  grad = compute_gradient(theta) # Tính grad = nabla J(theta^(t))
  # Điều kiện dừng sớm nếu đạo hàm đã triệt tiêu
  if abs(grad) < tolerance:
    print(f"Hội tụ tại bước {i}!") # Sửa lỗi biến 't' thành 'i'
    break

  theta = theta - learning_rate * grad # Cập nhật theta^(t+1)
  print(f"Bước {i+1}: theta = {theta:.4f}, Cost = {cost_function(theta):.4f}")
```

```python
# level 2
import numpy as np

# 1. Tạo dữ liệu giả lập: y = 2*x_1 + 3*x_2 + nhiễu
np.random.seed(42)
N, d = 100, 2
X = np.random.randn(N, d)
true_theta = np.array([[2.0], [3.0]])
y = X @ true_theta + np.random.randn(N, 1) * 0.05

# 2. Định nghĩa hàm Cost và Gradient bằng phép nhân ma trận
def compute_cost(X, y, theta):
    N = len(y)
    errors = X @ theta - y              # Kích thước: (N, 1)
    return (1 / (2 * N)) * np.sum(errors ** 2)

def compute_gradient(X, y, theta):
    N = len(y)
    errors = X @ theta - y              # Kích thước: (N, 1)
    # X^T @ errors: (d, N) @ (N, 1) -> (d, 1)
    return (1 / N) * (X.T @ errors)

# 3. Siêu tham số & Khởi tạo theta
learning_rate = 0.1
num_iterations = 60
theta = np.zeros((d, 1))                # Khởi tạo vector 0 kích thước (d, 1)

print(f"Bắt đầu: Cost = {compute_cost(X, y, theta):.4f}\n")

# 4. Vòng lặp Gradient Descent
for i in range(num_iterations):
    grad = compute_gradient(X, y, theta)
    theta = theta - learning_rate * grad

    if (i + 1) % 10 == 0:
        cost = compute_cost(X, y, theta)
        print(f"Bước {i+1:02d}: Cost = {cost:.6f} | theta = [{theta[0, 0]:.3f}, {theta[1, 0]:.3f}]")

print(f"\nKết quả tìm được: theta_1 ≈ {theta[0, 0]:.3f}, theta_2 ≈ {theta[1, 0]:.3f}")
```

```python
import torch

# 1. Tạo dữ liệu dạng Tensor (float32)
torch.manual_seed(42)
N, d = 100, 2
X = torch.randn(N, d)
true_theta = torch.tensor([[2.0], [3.0]])
y = X @ true_theta + torch.randn(N, 1) * 0.05

# 2. Khởi tạo tham số cần tối ưu (requires_grad=True)
theta = torch.zeros((d, 1), requires_grad=True)

learning_rate = 0.1
num_iterations = 60

# 3. Vòng lặp tối ưu
for i in range(num_iterations):
    # Forward pass: dự đoán và tính hàm loss
    y_hat = X @ theta
    loss = torch.mean((y_hat - y) ** 2) / 2

    # Backward pass: tự động tính nabla J(theta)
    loss.backward()

    # Cập nhật tham số (tắt tính toán đồ thị đạo hàm tạm thời)
    with torch.no_grad():
        theta -= learning_rate * theta.grad

        # Xóa gradient tích lũy sau mỗi bước lặp
        theta.grad.zero_()

    if (i + 1) % 10 == 0:
        print(f"Bước {i+1:02d}: Cost = {loss.item():.6f} | theta = [{theta[0, 0].item():.3f}, {theta[1, 0].item():.3f}]")
```

Exponential Moving Average (EMA): $v_t = \beta v_{t-1} + (1 - \beta) \theta_t$

```python
# 1. Dữ liệu đầu vào: chuỗi giá trị bị nhiễu rung lắc
raw_values = [10.0, 15.0, 12.0, 18.0, 14.0, 20.0, 16.0]

# 2. Siêu tham số (Hyperparameter)
beta = 0.9  # Hệ số ghi nhớ quá khứ

# 3. Khởi tạo biến trạng thái ban đầu v_0
v = raw_values[0]

print(f"Bắt đầu: v_0 = {v:.2f}\n")

# 4. Vòng lặp cập nhật lặp qua từng mốc thời gian t
for t, theta_t in enumerate(raw_values[1:], start=1):
    # Công thức: v_t = beta * v_(t-1) + (1 - beta) * theta_t
    v = beta * v + (1 - beta) * theta_t

    print(f"Bước {t}: Giá trị đo được theta_t = {theta_t:4.1f} --> EMA đã lọc v_t = {v:5.2f}")
```

Level 1 (Python thuần): Dùng vòng lặp for để tính 3 biến tích lũy: tích vô hướng, tổng bình phương của $u$, tổng bình phương của $v$. Không dùng math.dist hay thư viện ngoài (chỉ dùng math.sqrt).

Level 2 (NumPy): Viết trên 1 dòng duy nhất, dùng các toán tử @ (hoặc np.dot) và np.linalg.norm (hoặc np.sum).

Bài tập 2: Softmax ổn định số học (Numerically Stable Softmax)Chuyển đổi vector điểm số thực (logits) thành phân phối xác suất. Nếu tính trực tiếp $e^{z_i}$, số lớn như $z_i = 1000$ sẽ gây tràn số (inf). Do đó, cần trừ đi giá trị cực đại $M = \max(z)$ trước khi lấy mũ: $e^{z_i - M}$.Công thức toán:$$M = \max_{j=1}^K z_j, \qquad \sigma(z)_i = \frac{e^{z_i - M}}{\sum_{j=1}^K e^{z_j - M}}$$Yêu cầu:

Level 1 (Python thuần): Tìm max bằng vòng lặp, tính mảng các giá trị $e^{z_i - M}$ qua math.exp, tính tổng mẫu số, chia từng phần tử cho mẫu số.

Level 2 (NumPy xử lý Batch 2D): Mở rộng đầu vào thành ma trận kích thước $(N, K)$ gồm $N$ mẫu, mỗi mẫu có $K$ lớp. Áp dụng np.max và np.sum với tham số axis=-1, keepdims=True để chuẩn hóa song song toàn bộ ma trận mà không dùng vòng lặp.

```python
import math
```

```python
#Level 1
def softmax(z):
  M = max(z)
  total = 0.0
  for i in range(len(z)):
    exp_values = math.exp(z[i]-M)
    total += exp_values
  return exp_values/total
```

```python
#Level 2
def softmax_vec(z):
  M = np.max(z,axis=-1,keepdims=True)
  exp_values = np.exp(z-M)
  probabilities = exp_values / np.sum(exp_values,axis=-1,keepdims=True)
  return probabilities
```

Bài tập 3: Hàm mất mát Binary Cross-Entropy (BCE Loss)Đo sai lệch giữa phân phối nhãn thực tế $y \in \{0, 1\}$ và xác suất dự đoán $\hat{y} \in (0, 1)$.Công thức toán:$$\mathcal{L} = -\frac{1}{N} \sum_{i=1}^N \left[ y_i \ln(\hat{y}_i + \epsilon) + (1 - y_i) \ln(1 - \hat{y}_i + \epsilon) \right]$$(Với $\epsilon = 10^{-15}$ để tránh lỗi $\ln(0)$ khi $\hat{y} = 0$ hoặc $\hat{y} = 1$)Yêu cầu:

Level 1 (Python thuần): Dùng math.log, lặp qua từng cặp $(y_i, \hat{y}_i)$ với zip(), cộng dồn sai số vào biến total_loss, sau đó chia cho $N$ và đổi dấu.

Level 2 (NumPy / PyTorch): Vector hóa toàn bộ phép tính, xử lý cùng lúc mảng 1D hoặc 2D (N, 1) bằng các toán tử phần tử (*, -, np.log hoặc torch.log), lấy trung bình bằng np.mean() hoặc torch.mean().

```python
# Level 1
def BCE_loss(y_true,Y_pred):
  n = len(y_true)
  total = 0.0
  eps = 1e - 15
  for i in range(n):
    diff = y_true*math.log(y_pred + eps) + (1-y_true)*math.log(1-y_true +eps)
    total += diff
  return -total/n
```

```python
#Level 2
def BCE_loss_vec(y_true,Y_pred):
  eps = 1e-15
  return -np.mean(y_true*np.log(y_pred + eps) + (1-y_true)*np.log(1-y_pred + eps))
```

## 1. Two sum

```python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        # 1. Tạo danh sách chỉ số được sắp xếp theo giá trị của nums
        idx = sorted(range(len(nums)), key=lambda i: nums[i])

        # 2. Khởi tạo 2 con trỏ L và R
        L = 0
        R = len(nums) - 1

        # 3. Chạy khung 2 con trỏ
        while L < R:
            sum_val = nums[idx[L]] + nums[idx[R]]

            if sum_val == target:
                # FIX: Đảm bảo chỉ số trả về luôn được sắp xếp tăng dần [min, max]
                return sorted([idx[L], idx[R]])
            elif sum_val < target:
                L += 1
            else:
                R -= 1

        return []
```

## Contain duplicates

```python
class Solution:
    @staticmethod
    def hasDuplicate(arr: list[int]) -> bool:
        if not arr:
            return False

        # Bước 1: Bắt buộc phải sắp xếp mảng trước
        arr.sort()  # Độ phức tạp O(N log N)

        # Bước 2: Dùng 2 con trỏ kề nhau
        slow = 0
        for fast in range(1, len(arr)):
            # Nếu 2 ô kề nhau có giá trị bằng nhau -> Có trùng lặp
            if arr[slow] == arr[fast]:
                return True

            # Tăng slow lên để tiếp tục đi cùng fast
            slow += 1

        return False


# Test thử
nums = [1, 2, 3, 3]
print(Solution.hasDuplicate(nums))  # Kết quả: True
```
