EXAMPLE : $$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

```python
# Level 1 : raw loop
def calculate_mse(y, y_hat):
    n = len(y)
    total_error = 0.0
    for i in range(n):
        diff = y[i] - y_hat[i]
        total_error += diff ** 2
    return total_error / n
```

```python
# level 2 : Tư duy mảng (Vectorization - NumPy/Tensor)
import numpy as np

def calculate_mse_vec(y, y_hat):
    # Phép trừ và mũ hai thực hiện song song trên từng phần tử
    return np.mean((y - y_hat) ** 2)
```

```python
# euclidean level 1
def caculate_euclidean(u,v):
  n = len(u)
  total = 0.0
  for i in range(n):
    diff = u[i] - v[i]
    total += diff**2
  return np.sqrt(total)
```

```python
# manhattan level 1
def caculate_manhattan(u,v):
  n = len(u)
  total = 0.0
  for i in range(n):
    diff = u[i] - v[i]
    total += np.squrt(diff**2)
  return total
```

## 1. Two sum

```python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        # 1. Tạo danh sách chỉ số được sắp xếp theo giá trị của nums
        idx = sorted(range(len(nums)), key=lambda i: nums[i])

        # 2. Khởi tạo 2 con trỏ L và R
        L = 0
        R = len(nums) - 1

        # 3. Chạy khung 2 con trỏ
        while L < R:
            sum_val = nums[idx[L]] + nums[idx[R]]

            if sum_val == target:
                # FIX: Đảm bảo chỉ số trả về luôn được sắp xếp tăng dần [min, max]
                return sorted([idx[L], idx[R]])
            elif sum_val < target:
                L += 1
            else:
                R -= 1

        return []
```

## Contain duplicates

```python
class Solution:
    @staticmethod
    def hasDuplicate(arr: list[int]) -> bool:
        if not arr:
            return False

        # Bước 1: Bắt buộc phải sắp xếp mảng trước
        arr.sort()  # Độ phức tạp O(N log N)

        # Bước 2: Dùng 2 con trỏ kề nhau
        slow = 0
        for fast in range(1, len(arr)):
            # Nếu 2 ô kề nhau có giá trị bằng nhau -> Có trùng lặp
            if arr[slow] == arr[fast]:
                return True

            # Tăng slow lên để tiếp tục đi cùng fast
            slow += 1

        return False


# Test thử
nums = [1, 2, 3, 3]
print(Solution.hasDuplicate(nums))  # Kết quả: True
```
